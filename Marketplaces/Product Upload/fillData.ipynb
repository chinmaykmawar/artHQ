```python
def getColumns(fkfile, sheetname=2):
    cols = pd.read_excel(fkfile, sheet_name=sheetname).columns.to_list()
    columns= cols[cols.index("Seller SKU ID"):]
    if 'Unnamed: 8' in columns:
        columns.remove("Unnamed: 8")
    if 'Unnamed: 9' in columns:
        columns.remove("Unnamed: 9")
    columns.remove("Supplier Image")
    return columns
```

```python
def getCommonValues(pc):
    common_vals=product_vals[pc]
    for k,v in default_vals.items():
        if k not in common_vals.keys():
            common_vals[k]=v
    return common_vals
```

```python
def getUniqueValues(uvFile):
    unique_vals=pd.read_csv(uvFile, encoding='latin1')
    unique_vals.sort_index(key=lambda x: unique_vals["Seller SKU ID"], inplace=True)
    return unique_vals
    
```

```python
def getData(fkfile,pc,  uvfile):
    columns=getColumns(fkfile)
    common_vals=getCommonValues(pc)
    unique_vals=getUniqueValues(uvfile)
    
    df=pd.DataFrame(columns=columns)
    df["Seller SKU ID"]=product_to_be_Updated
    df.sort_index(key=lambda x: df["Seller SKU ID"], inplace=True)

    missing_cols=[]
    for col in df.columns:
        if col in unique_vals.columns:
            df.loc[:, col] = unique_vals.loc[:, col]
        elif col in common_vals.keys():
            df.loc[:, col] = common_vals[col]
        else:
            missing_cols.append(col)
    if len(missing_cols)==0:
        return df
    else:
        return missing_cols
```

```python
ProductCategory = "Set"
flipkart_FileName=r"C_jewellery-set_8c05f0197a4b40b4_2802-0259FK_REQQFXOQEITY2.xls"

fileDir = r"C:\Data\artHQ\Flipkart\\"
uniqueVals_FileName = "Unique_vals"
outputFileName = "Output_copy"

flipkart_FilePath = os.path.join(r"C:\Users\ArtHQ\Downloads", flipkart_FileName)
uniqueVals_FilePath = os.path.join(fileDir, uniqueVals_FileName+"_"+ProductCategory+".csv")
output_FilePath = os.path.join(fileDir, outputFileName+'.csv')
```

```python
data=getData(flipkart_FilePath, ProductCategory, uniqueVals_FilePath)

if not isinstance(data,pd.DataFrame):
    missing_cols=data
    cols=''
    for col in missing_cols:
        cols+=col+", "
    print(f"Missing Cols:{cols}")
elif data.isna().sum().sum()>0:
    cols=list(data.columns[data.isna().any()])
    #print(f"Missing Data:\n{data.columns[data.isna().any()]}")
    print(f"Missing Data:\n{data[cols]}")
else:
    final_df=pd.DataFrame(columns=data.columns)   
    for id in product_to_be_Updated:
        final_df=pd.concat([final_df, data[data["Seller SKU ID"]==id]])
    
    try:
        final_df.to_csv(output_FilePath, index=False, encoding='utf-8-sig')
    except PermissionError as e:
        print(f"Permission denied: {e}")
    else:
        print("File saved successfully at: "+output_FilePath)


    
```
